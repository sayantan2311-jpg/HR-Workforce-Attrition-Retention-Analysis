```python
import csv
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data=[]
with open(r"C:\Users\sulekha Jash\OneDrive\Desktop\Employee_data.csv","r",encoding="utf-8-sig") as f:
    reader=csv.DictReader(f)
    for row in reader:
        data.append(row)

sns.set_theme()
```

# Intermediate (Exploratory Data Analysis)

# Q11. Does overtime increase employee attrition?
Analysis: Bivariate
# Business Insight
Based on my analysis, I observed that employees who work overtime have a higher attrition rate than those who do not. This suggests that excessive workload and long working hours may increase stress and reduce employee satisfaction.
# Business Conclusion
I conclude that overtime is an important factor influencing employee attrition. Therefore, I recommend maintaining a healthy work-life balance and managing overtime effectively to improve employee retention.

```python
ot={"Yes":[0,0],"No":[0,0]}
for r in data:
    ot[r["Overtime"]][0]+=1
    if r["Attrition"]=="Left":
        ot[r["Overtime"]][1]+=1

x=["Yes","No"]
y=[ot[i][1]/ot[i][0] for i in x]
sns.barplot(x=x,y=y)
plt.title("Overtime vs Attrition")
plt.show()
```

# Q12. Is there a relationship betwwen Job Satisfaction and Attrition?
Analysis: Bivariate
# Business Insight
From my analysis, I found that employees with lower job satisfaction are more likely to leave the organization, while employees with higher satisfaction tend to stay longer.
# Conclusion
I conclude that improving job satisfaction through employee engagement, recognition, and career development can significantly reduce attrition.

```python
levels=["Low","Medium","High","Very High"]
rate=[]
for l in levels:
    t=a=0
    for r in data:
        if r["Job Satisfaction"]==l:
            t+=1
            if r["Attrition"]=="Left":
                a+=1
    rate.append(a/t)

sns.barplot(x=levels,y=rate)
plt.show()
```

# Q17. How does distance from home affect employee attrition?
## Business Insight 
Based on my analysis, I observed that employees who live farther from the workplace appear more likely to leave the organization. A longer commuting distance can increase travel time, cause stress and fatigue, and reduce work-life balance. # Q34. Create a heatmap showing correlations among numerical features.
Analysis: Multivariate
# Business Insight
Based on my correlation heatmap analysis, I observed the relationships among key numerical variables such as Age, Monthly Income, and Distance from Home. The heatmap makes it easy to identify whether the relationships are positive, negative, or weak. This visualization helps the organization understand how employee characteristics are connected and supports more informed business decisions.
# Business Conclusion 
I conclude that the correlation heatmap is a valuable analytical tool for identifying relationships among important employee variables. Although not every variable shows a strong correlation, the analysis provides meaningful insights that can support HR planning, compensation strategies, and employee retention decisions.

```python
age = []
income = []
distance = []

for r in data:
    try:
        a = float(r["Age"])
        i = float(r["Monthly Income"])
        d = float(r["Distance from Home"])

        age.append(a)
        income.append(i)
        distance.append(d)

    except ValueError:
        pass

arr = np.array([age, income, distance])

corr = np.corrcoef(arr)

sns.heatmap(
    corr,
    annot=True,
    cmap="YlGnBu",
    xticklabels=["Age","Income","Distance"],
    yticklabels=["Age","Income","Distance"]
)

plt.title("Correlation Heatmap")
plt.show()
```

# Q38. Visualize overtime versus attrition using a stacked bar chart
Analysis: Bivariate
# Business Insight
My analysis shows that employees who work overtime have a higher proportion of attrition compared to employees who do not work overtime. Excessive overtime may increase stress and reduce work-life balance.
# Business Conclusion
I conclude that managing overtime effectively can improve employee well-being, increase job satisfaction, and reduce employee turnover.

```python
yes_left=yes_stay=no_left=no_stay=0

for r in data:

    if r["Overtime"]=="Yes":
        if r["Attrition"]=="Left":
            yes_left+=1
        else:
            yes_stay+=1

    else:
        if r["Attrition"]=="Left":
            no_left+=1
        else:
            no_stay+=1

plt.bar(["Yes","No"],[yes_left,no_left],label="Left")
plt.bar(["Yes","No"],[yes_stay,no_stay],
        bottom=[yes_left,no_left],label="Stayed")

plt.legend()
plt.title("Overtime vs Attrition")
plt.show()
```

# Q10. What is the average performance rating by job level?
Analysis: Bivariate
# Business Insight
Based on my analysis, I compared the average performance rating across different job levels to understand employee performance trends.
# Business Conclusion
I conclude that evaluating performance ratings by job level helps management assess productivity, recognize high-performing employees, and support career development decisions.

```python
result = {}

for r in data:
    try:
        level = r["Job Level"]
        rating = float(r["Performance Rating"])

        if level not in result:
            result[level] = []

        result[level].append(rating)

    except:
        continue

for level, ratings in result.items():
    print("Job Level:", level)
    print("Average Performance Rating:", round(np.mean(ratings), 2))
```
